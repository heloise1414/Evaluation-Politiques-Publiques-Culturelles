## Statistiques descriptives

### Etablissements culturels en France

Dans un premier temps, nous allons regarder la répartition des établissements culturels sur le territoire. Le but ici est de faire une photographie du territoire afin de mieux comprendre les disparités qui peuvent exister quant à l'accès à la culture.

```python
carte_dep(df_final, "code_insee", "nb_total_etablissements", "Reds", "Nombre total d'établissements culturels en 2023")
```

A partir de cette première carte, plusieurs territoires se distinguent : Paris ressort clairement du lot avec une concentration d'équipements inégalée sur le reste du territoire. Les départements de la Gironde, du Nord et la région de la Bretagne se distinguent également, la Gironde et le Nord sont en effet parmi les départements étant le mieux équipés tandis que la Bretagne présente une homogénéité d'équipement entre les départements qui la composent.

En moyenne, les départements comptent 886,2 établissements et la médiane se situe à 763.

```python
carte_dep(df_final, "code_insee", "nb_libr", "Oranges", "Nombre de librairies en 2023")
```

Les librairies sont très concentrées dans certains territoires (à nouveau Paris). Cela se voit aussi bien sur la carte, que lorsque l'on constate que 75 % des départements disposent de moins de 56 librairies tandis que le maximum (Paris) est de 590 (soit dix fois plus).

```python
carte_dep(df_final, "code_insee", "nb_opera", "Oranges", "Nombre d'opéras en 2023")
```

Observer la répartition des opéras sur le territoire est particulièrement intéressant. En effet, moins de 25 % des départements disposent d'un opéra, et seule la ville de Paris en compte plusieurs. Les opéras sont en effet situés dans les grosses métropoles françaises.

Bien entendu, les spectateurs peuvent assister à des représentations d'opéras dans d'autres établissements (les théâtres notamment), mais la concentration de ce type de lieux dans certaines villes peut renforcer la distance de certains jeunes avec cet art, et renforcer leur sentiment d'illégitimité à son égard.

```python
carte_dep(df_final, "code_insee", "nb_ecrans_total", "Oranges", "Nombre de salles de cinéma en 2023")
```

Divertissement plus populaire que l'opéra ou le théâtre, le cinéma connaît lui aussi, malgré tout, une répartition inégale sur le territoire. Les équipements de cinéma semblent en effet être plus présents dans les départements les plus peuplés.
