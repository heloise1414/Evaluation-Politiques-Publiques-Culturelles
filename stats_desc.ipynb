### Etablissements culturels en France

Dans un premier temps, nous allons regarder la répartition des établissements culturels sur le territoire. Le but ici est de faire une photographie du territoire afin de mieux comprendre les disparités qui peuvent exister quant à l'accès à la culture.

```python
carte_dep(df_final, "code_insee", "nb_total_etablissements", "Reds", "Nombre total d'établissements culturels en 2023")
```

A partir de cette première carte, plusieurs territoires se distinguent : Paris ressort clairement du lot avec une concentration d'équipements inégalée sur le reste du territoire. Les départements de la Gironde, du Nord et la région de la Bretagne se distinguent également, la Gironde et le Nord sont en effet parmi les départements étant le mieux équipés tandis que la Bretagne présente une homogénéité d'équipement entre les départements qui la composent.

En moyenne, les départements comptent 886,2 établissements et la médiane se situe à 763.

```python
carte_dep(df_final, "code_insee", "nb_libr", "Oranges", "Nombre de librairies en 2023")
```

Les librairies sont très concentrées dans certains territoires (à nouveau Paris). Cela se voit aussi bien sur la carte, que lorsque l'on constate que 75 % des départements disposent de moins de 56 librairies tandis que le maximum (Paris) est de 590 (soit dix fois plus).

```python
carte_dep(df_final, "code_insee", "nb_opera", "Oranges", "Nombre d'opéras en 2023")
```

Observer la répartition des opéras sur le territoire est particulièrement intéressant. En effet, moins de 25 % des départements disposent d'un opéra, et seule la ville de Paris en compte plusieurs. Les opéras sont en effet situés dans les grosses métropoles françaises.

Bien entendu, les spectateurs peuvent assister à des représentations d'opéras dans d'autres établissements (les théâtres notamment), mais la concentration de ce type de lieux dans certaines villes peut renforcer la distance de certains jeunes avec cet art, et renforcer leur sentiment d'illégitimité à son égard.

```python
carte_dep(df_final, "code_insee", "nb_ecrans_total", "Oranges", "Nombre de salles de cinéma en 2023")
```

Divertissement plus populaire que l'opéra ou le théâtre, le cinéma connaît lui aussi, malgré tout, une répartition inégale sur le territoire. Les équipements de cinéma semblent en effet être plus présents dans les départements les plus peuplés.

### Matrice de corrélations

```python
# Liste des variables à inclure
variables = [
    'nb_biblio', 'nb_libr', 'nb_monum', 'nb_musee', 'nb_opera',
    'nb_theatre', 'nb_ecrans_total', 'nb_total_etablissements',
    'nb_jeunes', 'pop_total', 'rev_median', 'pauvrete',
    'aucun_diplome', 'dipl_sup'
]

# Calcul de la matrice de corrélation
corr = df_final[variables].corr()

# Création de la heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1, vmax=1, center=0,
    linewidths=0.5,
    square=True
)
plt.title('Matrice de corrélation', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

Cette matrice met en évidence plusieurs dynamiques structurantes. Tout d'abord, les variables liées aux équipements culturels sont fortement corrélées entre elles : nb_libr, nb_theatre, nb_ecrans_total et nb_opera présentent des coefficients compris entre 0,66 et 0,92, ce qui suggère que les territoires bien dotés en librairies tendent aussi à l'être en théâtres, cinémas et opéras — probablement un effet de taille urbaine ou de densité d'offre culturelle globale. Cette hypothèse est renforcée par les corrélations très élevées entre pop_total, nb_jeunes (0,94) et nb_total_etablissements (0,90), qui indiquent que la population totale est le principal moteur du nombre d'établissements culturels, plus que la nature spécifique de l'offre. La variable nb_monum se distingue en étant particulièrement liée à nb_total_etablissements (0,94), ce qui peut refléter un poids important des monuments historiques dans le total des établissements recensés.

Du côté socio-économique, on observe une opposition nette entre rev_median et pauvrete (-0,59), ainsi qu'entre dipl_sup et aucun_diplome (-0,86), cette dernière étant attendue puisque ces deux variables représentent les extrémités opposées de la distribution des niveaux de diplôme. Le niveau de revenu médian est par ailleurs fortement associé au niveau de diplôme (dipl_sup : 0,75 ; aucun_diplome : -0,75), confirmant le lien classique entre éducation et revenu. Il est intéressant de noter que ces variables socio-économiques sont globalement peu corrélées avec la présence d'équipements culturels bruts (coefficients souvent inférieurs à 0,5, voire proches de 0 pour pauvrete), à l'exception de rev_median qui montre une corrélation modérée avec nb_theatre (0,59) et nb_ecrans_total (0,70). Cela suggère que l'offre culturelle dépend davantage de la taille de la population que du niveau de richesse ou de pauvreté du territoire.
